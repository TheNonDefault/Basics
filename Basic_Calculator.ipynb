The following code demonstrates the code for a bascic calculator.

```python


def greet_user():
    """Greet the user, then ask for their name."""
    print("Hello mate!")
    name = input("What's your name? ").strip()
    if not name:
        name = "there"
    print(f"\nNice to meet you, {name}! Welcome to the Basic Calculator.\n")
    return name


def get_number(prompt):
    """Get a valid number from the user, re-prompting on invalid input."""
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("That doesn't look like a valid number. Please try again.")


def calculate(choice, num1, num2):
    """Perform the calculation based on the user's choice."""
    if choice == '1':
        return num1 + num2, '+'
    elif choice == '2':
        return num1 - num2, '-'
    elif choice == '3':
        return num1 * num2, '*'
    elif choice == '4':
        if num2 == 0:
            return None, '/'
        return num1 / num2, '/'


def format_num(n):
    """Display whole numbers without a trailing .0"""
    return int(n) if n.is_integer() else n


def main():
    name = greet_user()

    while True:
        print("What would you like to do?")
        print("  1. Addition (+)")
        print("  2. Subtraction (-)")
        print("  3. Multiplication (*)")
        print("  4. Division (/)")
        print("  5. Exit")

        choice = input("\nEnter your choice (1-5): ").strip()

        if choice == '5':
            print(f"\nThanks for using the calculator, {name}. Goodbye!")
            break

        if choice not in ('1', '2', '3', '4'):
            print("Invalid choice. Please enter a number between 1 and 5.\n")
            continue

        num1 = get_number("Enter the first number: ")
        num2 = get_number("Enter the second number: ")

        result, symbol = calculate(choice, num1, num2)

        if result is None:
            print("\nError: Division by zero is not allowed.\n")
        else:
            print(f"\nResult: {format_num(num1)} {symbol} {format_num(num2)} = {format_num(result)}\n")

        again = input("Would you like to do another calculation? (y/n): ").strip().lower()
        if again != 'y':
            print(f"\nThanks for using the calculator, {name}. Goodbye!")
            break


if __name__ == "__main__":
    main()
```
